# Raptor V5 + RadImageNet — Fusion v2 fine-tune

Branch `codex/raptor-v5-radimagenet-fusion`, training script `scripts/train_fusion_v2.py`.

* fp16 autocast + GradScaler (T4 has no bf16 tensor cores). BatchNorm statistics frozen.
* 1 studies per GPU per microstep; effective optimizer batch 16 studies. Gradient accumulation does not increase the BatchNorm batch.
* K=12 bags sampled evenly over present slots; series-boundary windows excluded.
* GPU augmentation: rotation/scale/shift, contrast/brightness/gamma, triplet-order reversal; no horizontal flip.
* Warmup + cosine LR (backbones 1e-5, fusion 1e-4, head 3e-5), EMA 0.998.
* Activation checkpointing follows RUN_CONFIG; the DDP memory probe uses the configured studies per GPU.
* Gold74 macro AUC (missing labels masked) and validation AUC/soft-BCE at step 0 (= parent Raptor)
  and after every epoch, for raw and EMA weights. With `fold` set, the validation set is that
  left-out fold and `fusion_best.pt` is selected by validation accuracy = mean of the gold74 and
  fold macro AUCs (`cv_metrics.csv`); otherwise by gold74 AUC.

```python
import json, os, subprocess, sys, torch
from pathlib import Path as _P
def _find(name):
    hits = sorted(p for base in ('/kaggle/input/notebooks', '/kaggle/input') for p in _P(base).glob(f'*/*/{name}') if 'rsna-fusion-ddp-train' in str(p))
    hits = hits or sorted(p for p in _P('/kaggle/input').glob(f'*/{name}') if 'rsna-fusion-ddp-train' in str(p))
    assert len(hits) == 1, (name, hits)
    return str(hits[0])
if RUN_CONFIG.get('resume_path') == 'auto':
    RUN_CONFIG['resume_path'] = _find('fusion_last.pt')
for key, value in list((RUN_CONFIG.get('extra_checkpoints') or {}).items()):
    if value.startswith('auto:'):
        RUN_CONFIG['extra_checkpoints'][key] = _find(value[5:])
print('RUN_CONFIG', RUN_CONFIG, flush=True)
run_folds = RUN_CONFIG.pop('run_folds', None)
runs = [(None, '/kaggle/working')] if run_folds is None else [(k, f'/kaggle/working/fold{k}') for k in run_folds]
env = dict(os.environ, PYTHONUNBUFFERED='1', OMP_NUM_THREADS='1',
           PYTORCH_CUDA_ALLOC_CONF='expandable_segments:True', PYTHONPATH=str(src))
for fold, workdir in runs:
    cfg = dict(RUN_CONFIG) if fold is None else {**RUN_CONFIG, 'fold': fold}
    os.makedirs(workdir, exist_ok=True)
    cmd = [sys.executable, '-m', 'torch.distributed.run', '--standalone',
           f'--nproc_per_node={torch.cuda.device_count()}', str(src/'train_fusion_ddp.py'),
           '--config', json.dumps(cfg)]
    print(' '.join(cmd), flush=True)
    proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True,
                            env=env, cwd=workdir, bufsize=1)
    for line in proc.stdout:
        print(line, end='', flush=True)
    if proc.wait() != 0:
        raise SystemExit(f'DDP training (fold {fold}) exited with code {proc.returncode}')
```
